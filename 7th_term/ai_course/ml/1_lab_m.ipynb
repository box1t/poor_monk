```python
# по книг, n_neighb = 1

class MyKNNСlf:
    
    def __init__(self, n_neighbors=1, metric="euclidean"):
        self.k = n_neighbors
        self.metric = metric

    def fit(self, X, y):
        # Просто запоминает данные
        self.X_train = np.array(X)
        self.y_train = np.array(y)
        return self

    def distance(self, a, b):
        if self.metric == "euclidean":
            return np.sqrt(np.sum((a - b) ** 2))
        elif self.metric == "manhattan":
            return np.sum(np.abs(a - b))

    def predict(self, X):
        X = np.array(X)
        predictions = []

        for x in X:
            distances = np.array([self.distance(x, train_x) for train_x in self.X_train])
            k_idx = distances.argsort()[:self.k]
            k_labels = self.y_train[k_idx]
            most_common = Counter(k_labels).most_common(1)[0][0]
            predictions.append(most_common)

        return np.array(predictions)

# используем улучшенный препроцессор
my_clf_pipe = Pipeline([
    ("prep", prep_clf_imp),
    ("model", MyKNNСlf(n_neighbors=3, metric="euclidean"))
])

my_clf_pipe.fit(Xclf_train, yclf_train)
yclf_pred_my = my_clf_pipe.predict(Xclf_test)

acc_my = accuracy_score(yclf_test, yclf_pred_my)
f1_my = f1_score(yclf_test, yclf_pred_my)

print(acc_my, f1_my)
```

```python
# своя регрессия

class MyKNNReg:
    def __init__(self, n_neighbors=3, metric="manhattan"):
        self.k = n_neighbors
        self.metric = metric

    def fit(self, X, y):
        self.X_train = np.array(X)
        self.y_train = np.array(y)
        return self

    def distance(self, a, b):
        if self.metric == "euclidean":
            return np.sqrt(np.sum((a - b) ** 2))
        elif self.metric == "manhattan":
            return np.sum(np.abs(a - b))

    def predict(self, X):
        X = np.array(X)
        preds = []

        for x in X:
            distances = np.array([self.distance(x, train_x) for train_x in self.X_train])
            k_idx = distances.argsort()[:self.k]
            preds.append(np.mean(self.y_train[k_idx]))

        return np.array(preds)

# используем улучшенный препроцессор
my_reg_pipe = Pipeline([
    ("prep", preprocessor_reg_imp),
    ("model", MyKNNReg(n_neighbors=3, metric="manhattan"))
])

my_reg_pipe.fit(Xreg_train, yreg_train)
yreg_pred_my = my_reg_pipe.predict(Xreg_test)

mae_my = mean_absolute_error(yreg_train, yreg_pred_my)
rmse_my = np.sqrt(mean_squared_error(yreg_train, yreg_pred_my))
r2_my = r2_score(yreg_train, yreg_pred_my)

print(mae_my, rmse_my, r2_my)
```
